```python
import pandas as pd
import os
import glob
```

```python
# Define the location of our raw data folder

raw_data_path = os.path.join("..", "Data", "Raw")

# Check whether the folder exists

print("Raw data folder exists:", os.path.exists(raw_data_path))
print("Raw data folder path:", raw_data_path)
```

```python
# Find all CSV files in the Raw folder

csv_files = glob.glob(os.path.join(raw_data_path, "*.csv"))

# Display the files

for file in csv_files:
    print(os.path.basename(file))
```

```python
# Create a dictionary to store our datasets

dataframes = {}

# Load every CSV file

for file in csv_files:
    dataset_name = os.path.basename(file).replace(".csv", "")
    dataframes[dataset_name] = pd.read_csv(file)

print("Datasets loaded:", len(dataframes))

# Display the dataset names
print(dataframes.keys())

# Display the first 5 rows of the orders dataset
dataframes["olist_orders_dataset"].head()
```

```python
# Inspect the dimensions of every dataset

for name, df in dataframes.items():
    print(f"{name}: {df.shape}")
```

```python
# Build a data inventory

inventory = []

for name, df in dataframes.items():
    inventory.append({
        "dataset_name": name,
        "row_count": df.shape[0],
        "column_count": df.shape[1],
        "memory_usage_mb": round(df.memory_usage(deep=True).sum() / (1024**2), 2)
    })

data_inventory = pd.DataFrame(inventory)

data_inventory
```

```python
# Sort datasets by row count, largest first

data_inventory = data_inventory.sort_values(
    by="row_count",
    ascending=False
)

data_inventory
```
